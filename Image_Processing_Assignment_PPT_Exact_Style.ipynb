## 5. Resize an image to 50% of original width and height

```python
import cv2
import matplotlib.pyplot as plt

img_cv = cv2.imread('flower.jpg')  # Read image
h, w = img_cv.shape[:2]
resized = cv2.resize(img_cv, (w // 2, h // 2))  # Resize to half
plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
plt.title('Resized Image')
plt.axis('off')
plt.show()
```

## 6. Crop the central 200 × 200 pixels and display using Matplotlib

```python
import cv2
import matplotlib.pyplot as plt

img_cv = cv2.imread('flower.jpg')  # Read image
h, w = img_cv.shape[:2]
x1 = (w - 200) // 2
y1 = (h - 200) // 2
cropped = img_cv[y1:y1 + 200, x1:x1 + 200]  # Crop centre
plt.imshow(cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB))
plt.title('Cropped Image')
plt.axis('off')
plt.show()
```

## 7. Rotate an image 90 degrees clockwise using OpenCV

```python
import cv2
import matplotlib.pyplot as plt

img_cv = cv2.imread('flower.jpg')  # Read image
rotated = cv2.rotate(img_cv, cv2.ROTATE_90_CLOCKWISE)  # Rotate clockwise
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.title('Rotated Image')
plt.axis('off')
plt.show()
```

## 8. Flip `car.jpg` left-to-right and display the result

```python
import cv2
import matplotlib.pyplot as plt

car = cv2.imread('car.jpg')  # Read image
flip_horizontal = cv2.flip(car, 1)  # Horizontal flip
plt.imshow(cv2.cvtColor(flip_horizontal, cv2.COLOR_BGR2RGB))
plt.title('Horizontal Flip')
plt.axis('off')
plt.show()
```

## 9. Convert image from RGB to HSV and display H, S, V channels

```python
import cv2
import matplotlib.pyplot as plt

img_cv = cv2.imread('flower.jpg')  # Read image in BGR
img_rgb = cv2.cvtColor(img_cv, cv2.COLOR_BGR2RGB)  # Convert BGR to RGB
hsv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)  # Convert RGB to HSV
h, s, v = cv2.split(hsv)  # Separate channels

plt.figure(figsize=(12, 3))
for i, (img, title) in enumerate(zip([h, s, v], ['Hue', 'Saturation', 'Value'])):
    plt.subplot(1, 3, i + 1)
    plt.imshow(img, cmap='gray')
    plt.title(title)
    plt.axis('off')
plt.show()
```

## 10. Calculate MSE between two grayscale images

```python
import cv2
from skimage.metrics import mean_squared_error

img1 = cv2.imread('img1.jpg', cv2.IMREAD_GRAYSCALE)  # Read first image
img2 = cv2.imread('img2.jpg', cv2.IMREAD_GRAYSCALE)  # Read second image
mse = mean_squared_error(img1, img2)  # Calculate error
print("Mean Squared Error (MSE):", mse)
```
